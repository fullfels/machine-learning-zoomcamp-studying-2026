```python
import pandas as pd
```

```python
file=pd.read_csv("https://raw.githubusercontent.com/alexeygrigorev/mlbookcamp-code/master/chapter-02-car-price/data.csv")
file
```

```python
file.columns = file.columns.str.lower().str.replace(" ", "_")
```

```python
file.dtypes
file.dtypes[file.dtypes == "str"]
```

```python
strings=list(file.dtypes[file.dtypes == "str"].index)
strings
```

```python
for col in strings:
    file[col]=file[col].str.lower().str.replace(" ", "_")
file.head()
```

```python
n=len(file)
n_test=int(n*0.2)
n_val=int(n*0.2)
n_train=n-n_test-n_val
n, n_val, n_test, n_train
```

```python
f_val = file.iloc[:n_val]
f_test = file.iloc[n_val:n_val+n_test]
f_train = file.iloc[n_val+n_test:]
```

```python
np.random.seed(2)

idx = np.arange(n)
np.random.shuffle(idx)
f_train = file.iloc[idx[:n_train]]
f_val = file.iloc[idx[n_train:n_train+n_val]]
f_test = file.iloc[idx[n_train+n_val:]]

```

```python
f_train = f_train.reset_index(drop=True)
f_val = f_val.reset_index(drop=True)
f_test = f_test.reset_index(drop=True)
```

```python
y_train = np.log1p(f_train.msrp.values)
y_val = np.log1p(f_val.msrp.values)
y_test = np.log1p(f_test.msrp.values)

del f_train['msrp']
del f_val['msrp']
del f_test['msrp']
```
